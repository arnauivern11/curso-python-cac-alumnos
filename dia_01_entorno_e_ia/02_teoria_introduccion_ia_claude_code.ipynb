## 1. Qué es Claude Code y en qué se diferencia de un chat genérico

Un chat genérico de IA (como una ventana de chat en un navegador) solo ve lo que le escribes en la conversación. No sabe qué archivos tiene tu proyecto, no puede abrirlos, y si le pegas código tienes que copiarlo y pegarlo tú a mano, en los dos sentidos.

**Claude Code** es un asistente que se ejecuta desde la terminal, dentro de la carpeta de tu proyecto, y tiene herramientas: puede **leer archivos, buscar texto en todo el repositorio, editar código, ejecutar comandos** (con tu permiso) y ver el resultado. La diferencia clave no es que "sea más listo": es que tiene **contexto real de tu proyecto** y puede **actuar** sobre él, no solo describir lo que habría que hacer.

Eso lo hace muy útil para programar, y también añade una responsabilidad: si puede editar archivos y ejecutar comandos, tienes que entender y revisar lo que hace, igual que revisarías el trabajo de alguien nuevo en el equipo antes de darlo por bueno.

## 3. El flujo básico

Cinco cosas que vas a hacer constantemente a partir de aquí. Pruébalas ahora mismo, en tu terminal, con Claude Code abierto en la carpeta del curso:

1. **Explorar el repositorio** — pídele que te oriente antes de tocar nada:
   > "Explícame la estructura de este repositorio: qué hay en cada carpeta de nivel superior."
2. **Explicar un notebook completo** — apunta a un archivo concreto:
   > "Abre `dia_01_entorno_e_ia/01_teoria_entorno_windows.ipynb` y resume en 5 líneas qué enseña."
3. **Explicar un fragmento de código** — pega o señala el fragmento exacto:
   > "En este notebook, explícame línea a línea qué hace la celda que usa `Path.cwd()` y `iterdir()`."
4. **Resolver un error sencillo** — pégale el mensaje de error completo, no un resumen tuyo:
   > "Al ejecutar esta celda me sale este error: `NameError: name 'raiz_curso' is not defined`. ¿Qué significa y cómo lo soluciono?"
5. **Generar documentación** — pídele un docstring o un README a partir de código ya escrito:
   > "Escribe un docstring en castellano para esta función, explicando qué hace, sus parámetros y qué devuelve."

Fíjate en el patrón: en los cinco casos le das algo **concreto** (un archivo, una celda, un mensaje de error literal), no una pregunta en el aire. Es la diferencia entre "no me funciona" y darle lo que necesita para ayudarte de verdad.

## 4. Anatomía de un buen prompt

Un buen prompt para programar tiene, casi siempre, estas cuatro partes:

| Parte | Qué aporta | Ejemplo |
|---|---|---|
| **Contexto** | Qué proyecto es, qué archivo, qué lenguaje/librería | "En este notebook de pandas, con el DataFrame `cartera`..." |
| **Objetivo** | Qué quieres conseguir, en concreto | "...quiero calcular la prima media por provincia..." |
| **Restricciones** | Qué no debe hacer, o qué debe respetar | "...usando `groupby`, no un bucle `for`, y sin modificar el DataFrame original..." |
| **Formato de salida** | Cómo quieres el resultado | "...devuélveme solo la línea de código, con un comentario de una línea explicando qué hace." |

Sin estas cuatro partes, el asistente tiene que **adivinar** lo que quieres — y cuando adivina, a veces acierta y a veces no. Con ellas, reduces la ambigüedad al mínimo.

### Ejemplo malo vs. bueno, sobre este mismo repositorio

**❌ Prompt malo:**

> "Arréglame el notebook de pandas."

Sin decir cuál notebook, qué está roto, ni qué "arreglado" significa para ti. El asistente tendría que explorar a ciegas y adivinar tu criterio de éxito.

**✅ Prompt bueno, con las cuatro partes:**

````python
prompt_bueno = """
Contexto: estoy en dia_07_pandas_numpy/03_ejercicios.ipynb, en el ejercicio 4,
que usa el dataset data/raw/compras_seguros.csv.

Objetivo: mi filtro `df[df["edad_cliente"] > 30]` me está devolviendo menos
filas de las que esperaba. Quiero entender por que y corregirlo.

Restricciones: no me des la solucion directamente. Antes, ayudame a
comprobar si hay valores nulos en la columna "edad_cliente" (se que ese
dataset los tiene a proposito, segun data/generators/README.md), porque
sospecho que ahi esta el motivo.

Formato de salida: primero una pregunta o comprobacion que deberia hacer yo
mismo, no la respuesta ya resuelta.
"""
print(prompt_bueno)
````

Con contexto (archivo y dataset concretos), objetivo (qué falla y qué se busca), restricción (que no dé la solución directa — pedagógicamente importante, ver `docs/uso_ia_en_el_curso.md`) y formato de salida (una pregunta guía, no la respuesta), Claude Code puede ayudarte de verdad sin hacer el ejercicio por ti.

**Para ti:** coge cualquier prompt que hayas escrito hace un momento en el paso 3 y reescríbelo con las cuatro partes explícitas, aunque el original ya funcionara. Verás que casi siempre se vuelve más corto de responder y más fácil de verificar.

## 5. Buenas prácticas y límites

Esto es lo que más importa de todo el notebook, y volverá en cada unidad del curso (ver `docs/uso_ia_en_el_curso.md` para la política completa):

- **No aceptes código que no entiendes.** Si Claude Code te da una línea con una función que no conoces, pregúntale qué hace antes de ejecutarla. "Cópialo y ya veremos" no es una estrategia, es una deuda.
- **Valida siempre.** Que el código se ejecute sin error no significa que haga lo correcto. Compara con un caso que ya sepas resolver a mano, o con un resultado de control.
- **No introduzcas datos sensibles.** Nunca pegues pólizas reales, datos de clientes o credenciales en un prompt, ni en este curso ni en tu trabajo real. Todos los datos de este curso son sintéticos precisamente para poder practicar sin ese riesgo.
- **La IA asiste, no decide.** Un LLM no tiene responsabilidad profesional sobre lo que firmas tú. Es una herramienta que acelera partes del trabajo; el criterio sigue siendo tuyo.
- **Las alucinaciones existen y son sutiles.** Un LLM puede generar código que se ve perfectamente razonable y aun así esté mal: llamar a una función que no existe, usar un argumento con el nombre equivocado, o afirmar un resultado que no ha calculado de verdad. No suelen "sonar" a error — por eso hace falta verificar, no solo leer con atención.

````python
# Ejemplo de alucinacion tipica: un nombre de metodo que "suena" a pandas
# pero no existe. Ejecuta esta celda para verlo con tus propios ojos.
import pandas as pd

df = pd.DataFrame({"producto": ["AUTO", "HOGAR", "AUTO"], "prima": [420, 260, 500]})

try:
    resultado = df.promedio_por_grupo("producto", "prima")  # no existe en pandas
except AttributeError as e:
    print("Esto es lo que pasaria si copiaras un codigo 'alucinado' sin comprobarlo:")
    print(type(e).__name__, "-", e)
````

Un LLM podría proponerte `df.promedio_por_grupo(...)` porque "suena" a algo que pandas debería tener, sin que exista de verdad. La única forma de detectarlo es **ejecutar el código** y, si funciona, **comprobar que el resultado es el que esperabas** — nunca dar por bueno un fragmento solo porque la explicación que lo acompaña suena convincente.
