## 1. Qué es un DataFrame

Un [`DataFrame`](https://pandas.pydata.org/docs/reference/frame.html) es el eje central de pandas: el equivalente a una hoja de cálculo de Excel o a una tabla de una base de datos. Es una estructura **bidimensional e indexada**: un conjunto de Series (columnas), potencialmente de tipos distintos entre sí, que comparten el mismo índice de filas.

En otras palabras: cada columna de un DataFrame es una `Series`, y todas comparten el mismo índice.

```python
import numpy as np
import pandas as pd
```

```python
# Un DataFrame se puede crear directamente desde un diccionario de listas
df_ejemplo = pd.DataFrame(
    data={
        "poliza_id": ["POL001", "POL002", "POL003"],
        "producto": ["AUTO", "HOGAR", "VIDA"],
        "prima_anual": [420.0, 310.5, 275.0],
    }
)
df_ejemplo
```

Fíjate en `fecha_alta` y `fecha_baja`: pandas las ha leído como `object` (texto), no como fecha, porque `read_csv()` no adivina tipos de fecha por defecto. Si necesitas operar con ellas como fechas de verdad (restar fechas, extraer el año...), hay que decírselo explícitamente con `parse_dates`:

```python
cartera_fechas = pd.read_csv(
    "../data/raw/cartera_polizas.csv",
    parse_dates=["fecha_alta", "fecha_baja"],
)
cartera_fechas.dtypes
```

> 📊 **En Excel esto sería...** como una celda que "parece" una fecha pero está formateada como texto: visualmente es indistinguible, pero no puedes restarle otra fecha ni usarla en un filtro de fechas hasta que cambias su formato de celda. `parse_dates` es el equivalente de decirle a pandas "formatea estas columnas como fecha".

## 4. Nombres de columnas: acceder, renombrar

Las columnas también son accesibles como atributo (`df.columna`), aunque la forma general y más segura es `df["columna"]` — lo verás en detalle en el próximo notebook. Para renombrarlas, hay dos vías principales.

### Modificar `.columns` directamente

```python
columnas_originales = cartera.columns.tolist()
print(columnas_originales[:4])

cartera_mayus = cartera.copy()
cartera_mayus.columns = [c.upper() for c in cartera_mayus.columns]
cartera_mayus.head(2)
```

Esto reemplaza **todos** los nombres de golpe, en el mismo orden — hay que tener cuidado de no desalinear nombres y columnas si se hace a mano.

### `.rename()`

Más flexible: permite cambiar solo algunas columnas, indicando explícitamente cuál va a cuál con un diccionario `{nombre_actual: nombre_nuevo}`.

```python
cartera.rename(columns={"poliza_id": "id_poliza", "prima_anual": "prima"}).head(2)
```

Por defecto, `.rename()` devuelve una **copia** con el cambio aplicado — el DataFrame original (`cartera`) no se modifica. Para que el cambio persista sobre la variable original hace falta `inplace=True`, o reasignar el resultado (`cartera = cartera.rename(...)`).

```python
print("¿Sigue igual el original?", "poliza_id" in cartera.columns)
```

También se puede renombrar de forma sistemática con una función, exactamente igual que con una *list comprehension* sobre `.columns`:

```python
cartera.rename(columns=str.upper).head(2)
```

> ⚠️ **Error típico**: escribir mal una clave del diccionario en `.rename(columns={...})` (una errata, mayúsculas distintas...). pandas **no avisa** si la clave no coincide con ninguna columna existente: simplemente no renombra nada y sigue sin error. Conviene comprobar el resultado (`.columns`) después de renombrar, no dar por hecho que ha funcionado.

```python
# "poliza_ID" con mayúscula no coincide con la columna real "poliza_id" -> no pasa nada, no hay error
resultado = cartera.rename(columns={"poliza_ID": "id_poliza"})
print("poliza_id" in resultado.columns, "id_poliza" in resultado.columns)
```

## Resumen

- Un DataFrame es una tabla bidimensional: un conjunto de Series (columnas) que comparten el mismo índice de filas — el equivalente de pandas a una hoja de cálculo.
- Se lee desde archivo con `pd.read_csv()` o `pd.read_excel()`; las rutas relativas dependen de desde dónde se ejecuta el notebook (`../data/raw/...` desde la carpeta del día).
- `read_csv()` no detecta fechas por defecto — hay que indicarlo con `parse_dates`, o esas columnas quedan como texto (`dtype object`) aunque "parezcan" fechas.
- Atributos básicos: `.shape` (filas, columnas), `.columns`, `.index`, `.dtypes` (revísalo siempre al cargar un dataset nuevo) y `.values` (la matriz de NumPy subyacente).
- Renombrar columnas: `.columns = [...]` sustituye todos los nombres de golpe; `.rename(columns={...})` cambia solo las indicadas y, por defecto, devuelve una copia (usa `inplace=True` o reasigna la variable para que persista).
- `.rename()` no avisa si una clave del diccionario no coincide con ninguna columna real — comprueba siempre el resultado.
