## 1. Qué es la iteración

**Iterar** es ejecutar el mismo bloque de código repetidamente. Hay dos formas:

- **Indefinida**: no se sabe de antemano cuántas veces se repetirá — se repite *mientras* se cumpla una condición. Es el bucle `while`, el protagonista de este notebook.
- **Definida**: se sabe de antemano sobre qué colección o cuántas veces se va a iterar. Es el bucle `for`, que veremos en el siguiente notebook.

## 2. El bucle `while`

```python
while <expresión>:
    <bloque>
```

Mientras `<expresión>` sea verdadera, `<bloque>` se ejecuta una y otra vez. Python la revisa **antes** de cada repetición: en cuanto es `False`, el bucle termina.

> 📊 **En Excel esto sería...** no tiene un equivalente directo de fórmula (Excel no itera "mientras" algo sea cierto salvo con macros de VBA). Es más parecido a repetir manualmente "Actualizar → comprobar → Actualizar → comprobar..." hasta que una celda de control cambia de valor.

````python
# Simulamos el pago de una prima fraccionada: se van descontando cuotas
# de un capital pendiente hasta saldarlo.
capital_pendiente = 1200.0
cuota = 300.0

while capital_pendiente > 0:
    print(f"Capital pendiente: {capital_pendiente:.2f} € -> se paga una cuota de {cuota:.2f} €")
    capital_pendiente -= cuota

print(f"Capital pendiente final: {capital_pendiente:.2f} € (póliza saldada)")
````

Fíjate en `capital_pendiente -= cuota`: es lo que hace avanzar la condición del bucle. **Sin esa línea, `capital_pendiente` nunca cambiaría** y el bucle no terminaría jamás — es exactamente el error que veremos más abajo.

## 3. `break` y `continue`

- `break`: corta el bucle **por completo**, de inmediato.
- `continue`: salta el resto del bloque y pasa a la **siguiente** comprobación de la condición, sin cortar el bucle.

````python
# break: dejamos de procesar la cola de siniestros en cuanto encontramos uno fraudulento
cola_siniestros = ["S001", "S002", "FRAUDE-S003", "S004", "S005"]
indice = 0

while indice < len(cola_siniestros):
    siniestro = cola_siniestros[indice]
    if siniestro.startswith("FRAUDE"):
        print(f"Siniestro sospechoso detectado ({siniestro}): se detiene el procesamiento automático.")
        break
    print(f"Procesando {siniestro}...")
    indice += 1
````

````python
# continue: saltamos los siniestros ya cerrados, sin cortar el resto de la cola
siniestros_estado = {"S001": "abierto", "S002": "cerrado", "S003": "abierto", "S004": "cerrado"}
ids = list(siniestros_estado.keys())
i = 0

while i < len(ids):
    id_siniestro = ids[i]
    i += 1  # avanzamos SIEMPRE, esté cerrado o no -- ver aviso más abajo
    if siniestros_estado[id_siniestro] == "cerrado":
        continue
    print(f"Siniestro abierto pendiente de revisión: {id_siniestro}")
````

> ⚠️ **Error típico: el bucle infinito**
>
> El error más costoso con `while` no es uno que Python detecte por ti: no hay traceback, el kernel simplemente **se queda colgado**. Pasa cuando la condición del bucle nunca deja de ser verdadera, casi siempre porque se olvida actualizar la variable de control.
>
> **No vamos a ejecutar el ejemplo real** (colgaría este notebook y con `nbconvert` nunca terminaría de ejecutarse). En su lugar, lo mostramos como texto y lo razonamos:
>
> ```python
> i = 0
> while i < 5:
>     print(i)
>     # falta "i += 1" aquí -- i nunca cambia, la condición i < 5 es
>     # siempre verdadera, el bucle no termina nunca
> ```
>
> Si esto te pasara de verdad en tu propio notebook, la señal es un asterisco `[*]` que no cambia nunca en el margen izquierdo de la celda, y la única salida es **interrumpir el kernel** (el botón de "stop" ⏹ en la barra de herramientas, o `Kernel → Interrupt`), corregir el código, y volver a ejecutar.

````python
# Versión segura para demostrar el mismo patrón sin arriesgarnos a colgar el notebook:
# añadimos un límite de seguridad (un "cortafuegos") que nunca deberías necesitar
# en un bucle bien escrito, pero que aquí usamos solo para poder mostrar el fallo sin peligro.
i = 0
limite_de_seguridad = 5
iteraciones = 0

while i < limite_de_seguridad:
    print(f"i sigue siendo {i} -- si esto fuera un bucle real sin actualizar 'i', no pararía nunca")
    iteraciones += 1
    if iteraciones >= 3:
        print("(cortamos aquí a propósito para no generar una salida interminable)")
        break
    # OJO: en un bucle real, aquí faltaría "i += 1"
````

## 4. La cláusula `else` de un `while`

Python permite un `else` opcional al final de un `while`. Se ejecuta **solo si el bucle termina de forma natural** (la condición se vuelve `False`), **no** si termina por un `break`. Es poco común, pero útil para distinguir "terminé normalmente" de "corté antes de tiempo".

````python
# Revisamos siniestros buscando uno con importe superior al límite de la póliza;
# si no lo encontramos en ninguno, el "else" confirma que la cartera está limpia.
importes_siniestros = [1200, 3400, 800, 2100]
limite_poliza = 10_000

i = 0
while i < len(importes_siniestros):
    if importes_siniestros[i] > limite_poliza:
        print(f"Siniestro {i} supera el límite de póliza: revisar manualmente.")
        break
    i += 1
else:
    print("Revisión completa: ningún siniestro supera el límite de la póliza.")
````

## 5. `input()` y cómo lo usamos en este curso

`input()` muestra un mensaje y **detiene la ejecución** hasta que la persona usuaria escribe algo y pulsa Enter. Es habitual combinarlo con `while` para pedir datos hasta que sean válidos.

> ⚠️ **Por qué no usamos `input()` de verdad en los notebooks del curso**: cuando un notebook se ejecuta de forma automática (por ejemplo, con `jupyter nbconvert --execute`, que es como se valida cada notebook de este curso), no hay ninguna persona delante para escribir una respuesta — el notebook se queda esperando para siempre y la ejecución nunca termina. Por eso, en el material de este curso **simulamos** la entrada del usuario con una variable ya asignada, y lo dejamos explícito en el texto. En un uso real, fuera de este notebook, esa misma línea sería `entrada_usuario = input("...")`.

````python
# En un script o notebook de uso real (ejecutado por una persona, no de forma automática),
# esta línea sería:
#     entrada_usuario = input("Escribe el número de póliza a consultar: ")
# Aquí SIMULAMOS lo que escribiría el usuario con una variable ya asignada,
# para que el notebook se pueda ejecutar de principio a fin sin intervención manual.
entrada_usuario = "P002"  # <-- simula lo que teclearía la persona usuaria

if entrada_usuario in [p["id"] for p in [
    {"id": "P001"}, {"id": "P002"}, {"id": "P003"}, {"id": "P004"}, {"id": "P005"},
]]:
    print(f"Póliza {entrada_usuario} encontrada.")
else:
    print(f"Póliza {entrada_usuario} no existe en la cartera.")
````

Un patrón muy habitual con `while` + `input()` es repetir la pregunta hasta obtener una respuesta válida. En un script real:

```python
while True:
    entrada_usuario = input("Escribe 'fin' para terminar: ")
    if entrada_usuario == "fin":
        print("Finalizando.")
        break
```

Para que este mismo patrón se pueda ejecutar sin intervención manual, lo simulamos con una **lista de respuestas ya preparada** que vamos consumiendo, como si fuera lo que la persona usuaria habría ido escribiendo en cada vuelta del bucle.

````python
# Simulación de varias llamadas a input() consecutivas: en vez de detenernos a esperar
# tecleo real, consumimos una lista de respuestas ya decididas de antemano.
respuestas_simuladas = ["P009", "abc", "P002", "fin"]  # lo que "escribiría" el usuario, en orden
turno = 0

while True:
    # entrada_usuario = input("Escribe un ID de póliza o 'fin' para terminar: ")  # uso real
    entrada_usuario = respuestas_simuladas[turno]  # uso simulado, para poder ejecutar sin manos
    turno += 1

    if entrada_usuario == "fin":
        print("Finalizando consulta.")
        break
    elif not entrada_usuario.startswith("P"):
        print(f"'{entrada_usuario}' no parece un ID de póliza válido (debe empezar por 'P').")
        continue
    else:
        print(f"Consultando póliza {entrada_usuario}...")
````

**Para ti:** cambia `respuestas_simuladas` para que la primera respuesta ya sea `"fin"`. Antes de ejecutar, predice cuántas líneas imprimirá el bucle. Ejecuta y comprueba si acertaste.

## Resumen

- `while <expr>:` repite un bloque mientras `<expr>` sea verdadera; Python la revisa antes de cada vuelta.
- `break` corta el bucle por completo; `continue` salta a la siguiente comprobación de la condición sin cortar el bucle.
- El `else` de un `while` se ejecuta solo si el bucle termina de forma natural (condición `False`), no si termina con `break` — útil para distinguir "no encontré nada raro" de "corté porque encontré algo".
- El bucle infinito es el error más costoso de un `while`: pasa cuando se olvida actualizar la variable que controla la condición. No da traceback — el kernel se queda colgado (`[*]` fijo) y hay que interrumpirlo a mano.
- `input()` bloquea la ejecución esperando tecleo real; en un notebook que se ejecuta de forma automática (como los de este curso, validados con `nbconvert --execute`) eso cuelga el proceso para siempre.
- Por eso, en este curso simulamos `input()` con una variable (o lista de variables) ya asignada, dejando siempre explícito en el texto cuál sería la línea real con `input()`.
