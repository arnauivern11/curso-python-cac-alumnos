## 1. Qué es una tupla

Una **tupla** es una colección ordenada, muy parecida a una lista, con una diferencia clave: es **inmutable**. Se crea con paréntesis `()`, aunque en realidad lo que define una tupla son las **comas**, no los paréntesis.

```python
cobertura = (1, 2, "responsabilidad_civil", "lunas")
print(cobertura, type(cobertura))

# tambien se puede construir con la funcion tuple()
otra_tupla = tuple([1, 2, "a", "b"])
print(otra_tupla)

# los parentesis son opcionales: lo que crea la tupla es la coma
coordenadas_riesgo = 41.38, 2.17   # latitud, longitud de un riesgo asegurado
print(coordenadas_riesgo, type(coordenadas_riesgo))
```

## 2. Inmutabilidad

Al igual que un string, una tupla no permite modificar sus elementos una vez creada.

```python
registro_poliza = ("AUTO-2026-00147", "2026-01-15", 480.5)
print(registro_poliza[0])

try:
    registro_poliza[2] = 500.0
except TypeError as e:
    print("Esto pasa si intentas modificar un elemento de una tupla:")
    print(type(e).__name__, "-", e)
```

**Matiz importante:** la inmutabilidad de la tupla es sobre **qué elementos contiene**, no sobre si esos elementos son a su vez mutables. Si una tupla contiene una lista, esa lista sigue siendo mutable — puedes modificar su contenido, aunque no puedas reemplazarla por otra lista distinta dentro de la tupla.

```python
historial_siniestros = ([1200, 850], [0, 3400])   # una tupla de dos listas

historial_siniestros[0][1] = 900   # SI se puede: modificas la LISTA que hay dentro, no la tupla en si
print(historial_siniestros)

try:
    historial_siniestros[0] = [1, 2]   # esto NO se puede: reemplazar un elemento de la tupla
except TypeError as e:
    print("Esto pasa si intentas reemplazar un elemento (no su contenido) de la tupla:")
    print(type(e).__name__, "-", e)
```

## 3. Métodos y *unpacking*

```python
coberturas = ("responsabilidad_civil", "lunas", "asistencia", "responsabilidad_civil")

print(coberturas.index("lunas"))                    # posicion de la PRIMERA aparicion
print(coberturas.count("responsabilidad_civil"))     # cuantas veces aparece
```

El ***unpacking*** (desempaquetado) es una forma muy común de usar tuplas: asignar cada elemento a una variable distinta en una sola línea. Lo verás constantemente a partir de aquí — por ejemplo, cuando una función devuelva varios valores a la vez (Día 4).

```python
registro_poliza = ("AUTO-2026-00147", "2026-01-15", 480.5)

poliza_id, fecha_inicio, prima = registro_poliza   # unpacking: una variable por posicion
print(poliza_id)
print(fecha_inicio)
print(prima)
```

## 4. ¿Cuándo usar tuplas?

Con listas y tuplas pudiendo hacer cosas parecidas, la pregunta razonable es cuándo elegir cada una. Algunos criterios prácticos:

- **Datos que no deben cambiar una vez creados.** Un registro de siniestro `(siniestro_id, poliza_id, fecha)` no debería reescribirse por accidente en medio de un cálculo — una tupla te protege de ese bug (`TypeError` si lo intentas, en vez de un dato corrompido en silencio).
- **Como clave de un diccionario.** Las claves de un `dict` deben ser inmutables — una tupla **sí puede** ser clave, una lista **no puede**. Es útil para indexar por combinaciones, como `(poliza_id, anio)` para localizar la prima de una póliza en un año concreto.
- **Cuando una función devuelve varios valores.** En Python, `return a, b` empaqueta `a` y `b` en una tupla automáticamente — lo verás mucho a partir del Día 4.
- **Ligereza.** Una tupla ocupa algo menos de memoria y es marginalmente más rápida de crear que una lista equivalente — rara vez es el motivo principal para elegirla, pero es una ventaja añadida cuando ya tenía sentido por inmutabilidad.

Como regla rápida: si vas a **construir la colección poco a poco** (añadiendo, quitando, reordenando), usa una lista. Si representa **un registro fijo que ya está completo** cuando lo creas, una tupla dice mejor esa intención — y de paso evita que alguien (tú mismo dentro de tres meses) lo modifique sin querer.

```python
# Ejemplo: usar una tupla (poliza_id, anio) como clave de diccionario
primas_por_poliza_y_anio = {
    ("AUTO-2026-00147", 2025): 460.0,
    ("AUTO-2026-00147", 2026): 480.5,
    ("HOGAR-2026-00033", 2026): 275.0,
}

print(primas_por_poliza_y_anio[("AUTO-2026-00147", 2026)])

try:
    # una lista NO puede ser clave de diccionario (no es "hashable")
    primas_por_poliza_y_anio[["AUTO-2026-00147", 2026]]
except TypeError as e:
    print("Esto pasa si intentas usar una lista como clave de diccionario:")
    print(type(e).__name__, "-", e)
```

## Resumen

- Una tupla se crea con `()` (o solo con comas); es como una lista pero **inmutable**.
- Intentar modificar un elemento de una tupla da `TypeError` — es la protección que ofrece frente a una lista.
- Si una tupla contiene un elemento mutable (como una lista), ese elemento interno **sí se puede modificar** — la inmutabilidad es sobre qué elementos tiene la tupla, no sobre el contenido de esos elementos.
- `.index()` y `.count()` funcionan igual que en listas.
- El *unpacking* (`a, b, c = tupla`) es el uso más frecuente de las tuplas en código real.
- Usa tuplas para registros fijos que no deben cambiar, como claves de diccionario (`(poliza_id, anio)`) o para los valores que devuelve una función.
