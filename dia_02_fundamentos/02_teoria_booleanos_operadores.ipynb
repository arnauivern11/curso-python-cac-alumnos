## 1. El tipo `bool`

`bool` es el tipo de dato más básico: solo puede valer `True` o `False`. Internamente, `True` equivale a `1` y `False` a `0` (por eso puedes sumarlos, aunque rara vez tenga sentido hacerlo).

```python
poliza_activa = True
poliza_siniestrada = False

print(poliza_activa, type(poliza_activa))
print(poliza_activa + poliza_activa)  # True + True == 2, aunque casi nunca se use asi
```

## 2. Operadores de comparación

| Operador | Significado |
|---|---|
| `==` | igual a |
| `!=` | distinto de |
| `<` / `>` | menor / mayor que |
| `<=` / `>=` | menor o igual / mayor o igual |

El resultado de comparar siempre es un `bool`.

```python
prima_actual = 420
prima_limite_riesgo = 500

print(prima_actual < prima_limite_riesgo)   # ¿esta por debajo del limite de riesgo?
print(prima_actual == 420)
print(prima_actual != prima_limite_riesgo)
```

> 📊 **En Excel esto sería...** el resultado que da una celda con una fórmula como `=B2<500`: Excel también te devuelve `VERDADERO`/`FALSO`. La diferencia es que en Python ese resultado es un valor (`bool`) que puedes guardar en una variable, combinarlo con otros y reutilizarlo, no solo mostrarlo en una celda.

> ⚠️ **Error típico**: escribir `prima_actual = 500` cuando querías **comparar** (`==`) en vez de **asignar** (`=`). `=` con un solo signo cambia el valor de la variable; `==` con dos signos compara sin modificar nada. Es uno de los errores más comunes al empezar, y Python no avisa porque `prima_actual = 500` es una instrucción perfectamente válida, solo que no es la que querías.

## 3. Operadores lógicos: `and`, `or`, `not`

Sirven para combinar varias condiciones en una sola expresión booleana.

```python
antiguedad_anios = 3
siniestros_ultimo_anio = 0

# Una poliza es "bonificable" si lleva mas de 2 anios y no ha tenido siniestros
es_bonificable = (antiguedad_anios > 2) and (siniestros_ultimo_anio == 0)
print("Bonificable:", es_bonificable)

# Es "revisable" si es muy nueva O ha tenido siniestros
es_revisable = (antiguedad_anios < 1) or (siniestros_ultimo_anio > 0)
print("Revisable:", es_revisable)

print("No bonificable:", not es_bonificable)
```

`and` y `or` son de **cortocircuito** (*short-circuit*): en `A and B`, si `A` ya es `False`, Python ni siquiera evalúa `B` porque el resultado ya está decidido. Es relevante cuando `B` es una operación costosa, o cuando `B` fallaría si `A` no se cumpliera primero (por ejemplo, comprobar que una lista no está vacía antes de acceder a su primer elemento).

> 📊 **En Excel esto sería...** las funciones `Y()` (`AND`) y `O()` (`OR`): `=Y(B2>2; C2=0)` es exactamente `(antiguedad_anios > 2) and (siniestros_ultimo_anio == 0)`. La lógica es idéntica; solo cambia la sintaxis.

**Para ti:** una póliza de auto es candidata a "renovación automática" si `antiguedad_anios >= 1` y `prima_actual <= 600`, o si es un cliente `es_vip`. Escribe la expresión booleana completa con las variables de la celda siguiente.

```python
antiguedad_anios = 2
prima_actual = 480
es_vip = False

# tu codigo aqui
candidata_renovacion_automatica = (antiguedad_anios >= 1 and prima_actual <= 600) or es_vip
print(candidata_renovacion_automatica)
```

## 4. `==` frente a `is`

`==` compara si dos valores **son iguales**. `is` compara si dos nombres apuntan **al mismo objeto en memoria** (mismo `id()`). Casi siempre quieres `==`; `is` se reserva para comparar con `None` o para comprobar si dos variables son literalmente el mismo objeto.

```python
cartera_a = [1, 2, 3]
cartera_b = cartera_a.copy()   # copia el contenido, pero es un objeto NUEVO

print("¿Mismo contenido? ", cartera_a == cartera_b)   # True: mismos valores
print("¿Mismo objeto?    ", cartera_a is cartera_b)   # False: objetos distintos en memoria
print("id cartera_a:", id(cartera_a), "| id cartera_b:", id(cartera_b))
```

```python
cartera_c = cartera_a   # NO copia: cartera_c es otro NOMBRE para el mismo objeto

print("¿Mismo objeto cartera_a y cartera_c?", cartera_a is cartera_c)
```

> ⚠️ **Error típico**: usar `is` para comparar números o texto (`prima is 420`) esperando que se comporte como `==`. A veces "funciona por casualidad" (Python reutiliza internamente algunos enteros pequeños), pero no es su uso correcto y puede fallar de forma inconsistente según el valor. Para comparar valores, usa siempre `==`; usa `is` solo con `None` (`if resultado is None:`) o cuando de verdad te interese la identidad del objeto, no su contenido.

## 5. Bonus: `set` — valores únicos

Un `set` es una colección **no ordenada de elementos únicos**: si metes un valor repetido, solo se queda una copia. No forma parte estrictamente del programa de esta unidad, pero es una herramienta pequeña y muy útil para comparar listados — por ejemplo, qué clientes tienen pólizas en dos ramos distintos.

```python
clientes_auto = {"C001", "C002", "C003", "C004"}
clientes_hogar = {"C002", "C003", "C005"}

print("En ambos ramos (interseccion):", clientes_auto & clientes_hogar)
print("En cualquiera de los dos (union):", clientes_auto | clientes_hogar)
print("Solo en auto (diferencia):", clientes_auto - clientes_hogar)
print("Solo en uno de los dos, no en ambos (dif. simetrica):", clientes_auto ^ clientes_hogar)
print("¿C002 tiene auto?", "C002" in clientes_auto)
```

> 📊 **En Excel esto sería...** parecido a comparar dos columnas con `Quitar duplicados` combinado con `COINCIDIR`: la intersección (`&`) son los valores presentes en ambas columnas, la unión (`|`) es el listado combinado sin duplicados, y la diferencia (`-`) son los que están en una pero no en la otra — cosas que en Excel suelen requerir varias fórmulas o una tabla dinámica, y aquí son un solo operador.

**Para ti:** dado `polizas_impagadas = {"P010", "P022", "P031"}` y `polizas_siniestradas = {"P022", "P045"}`, calcula qué pólizas están impagadas **y también** siniestradas (intersección), y cuántas pólizas distintas hay en total entre ambos listados (longitud de la unión).

```python
polizas_impagadas = {"P010", "P022", "P031"}
polizas_siniestradas = {"P022", "P045"}

# tu codigo aqui
impagadas_y_siniestradas = polizas_impagadas & polizas_siniestradas
total_polizas_distintas = len(polizas_impagadas | polizas_siniestradas)

print(impagadas_y_siniestradas)
print(total_polizas_distintas)
```

## Resumen

- `bool` solo puede valer `True` o `False`; comparar con `==`, `!=`, `<`, `>`, `<=`, `>=` siempre devuelve un `bool`.
- `and`/`or`/`not` combinan condiciones; `and`/`or` evalúan de izquierda a derecha y paran en cuanto el resultado ya está decidido (*short-circuit*).
- `=` asigna, `==` compara — confundirlos es uno de los errores más comunes y silenciosos al empezar.
- `==` compara valores; `is` compara identidad de objeto (mismo `id()`). Usa `==` salvo que necesites explícitamente lo segundo (o compares con `None`).
- `set` guarda valores únicos y soporta unión (`|`), intersección (`&`), diferencia (`-`) y diferencia simétrica (`^`) — muy útil para comparar listados de pólizas o clientes.
