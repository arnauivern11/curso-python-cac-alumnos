## 1. El bucle `for`

A diferencia de `while` (iteración indefinida, "mientras se cumpla algo"), `for` es iteración **definida**: recorre, uno a uno y en orden, los elementos de una colección.

```python
for elemento in <coleccion>:
    <bloque>
```

> 📊 **En Excel esto sería...** una fórmula que se copia hacia abajo por cada fila de una tabla: la misma operación, aplicada elemento a elemento, sin que tú repitas nada a mano.

````python
cartera_auto = ["P001", "P004", "P007", "P009"]

for id_poliza in cartera_auto:
    print(f"Revisando póliza de auto: {id_poliza}")
````

### Iterables e iteradores

Un objeto es **iterable** si se puede recorrer elemento a elemento (listas, tuplas, strings, sets, diccionarios...). Internamente, `for` le pide a ese objeto un **iterador** con la función `iter()`, y va llamando a `next()` sobre él hasta que se agota.

````python
print("Iterador de una lista:  ", iter([1, 2, 3]))
print("Iterador de una cadena:  ", iter("auto"))
print("Iterador de una tupla:   ", iter((1, 2, 3)))
print("Iterador de un set:      ", iter({1, 2, 3}))
print("Iterador de un dict:     ", iter({"a": 1, "b": 2}))
````

````python
primas_recientes = [480.50, 210.00, 650.00]
iterador = iter(primas_recientes)

print(next(iterador))
print(next(iterador))
print(next(iterador))
````

> ⚠️ **Error típico: `StopIteration`**
>
> Un iterador solo avanza hacia adelante y solo tiene los elementos que tenía la colección original. Si le pides un valor más de los que quedan, lanza `StopIteration`. Normalmente nunca ves este error de forma directa porque `for` lo gestiona por ti automáticamente (para él **es** la señal de "ya he terminado, salgo del bucle") — pero si usas `next()` a mano, como aquí, sí puede aparecer.

````python
try:
    next(iterador)  # ya hemos consumido los 3 elementos de "primas_recientes"
except StopIteration as e:
    print(f"{type(e).__name__}: no quedan más elementos en el iterador.")
````

## 2. Iteración en cadenas de texto

Las cadenas son iterables: `for` recorre carácter a carácter.

````python
for letra in "auto":
    print(letra.upper(), end=" ")
print()
````

## 3. Iteración en diccionarios

Por defecto, iterar sobre un diccionario recorre sus **llaves**. Para llaves, valores, o ambos a la vez, están `.keys()`, `.values()` y `.items()`.

````python
primas_por_producto = {"auto": 480.50, "hogar": 210.00, "vida": 650.00}

print("Por defecto (llaves):")
for producto in primas_por_producto:
    print(" -", producto)

print("\nCon .values():")
for prima in primas_por_producto.values():
    print(" -", prima)

print("\nCon .items() (llave y valor a la vez, con tuple unpacking):")
for producto, prima in primas_por_producto.items():
    print(f" - {producto}: {prima} €")
````

> 📊 **En Excel esto sería...** recorrer dos columnas emparejadas (por ejemplo "Producto" y "Prima") fila a fila a la vez, en vez de mirar cada columna por separado.

## 4. `range()`: generar secuencias numéricas sin escribirlas a mano

````python
print(list(range(5)))          # 0, 1, 2, 3, 4  (empieza en 0, no incluye el 5)
print(list(range(1, 5)))       # 1, 2, 3, 4      (empieza en 1)
print(list(range(0, 10, 2)))   # de 2 en 2
````

````python
# Simulamos 5 años de una póliza de vida con una prima que crece un 3% anual
prima_inicial = 650.00

for anio in range(1, 6):
    prima_anio = prima_inicial * (1.03 ** (anio - 1))
    print(f"Año {anio}: prima = {prima_anio:.2f} €")
````

> ⚠️ **Error típico: confundir el límite de `range()`**
>
> `range(1, 6)` genera `1, 2, 3, 4, 5` — **no incluye el 6**. Es un error de lógica muy común (no da ningún traceback, simplemente el resultado tiene un elemento de menos o de más de lo esperado) esperar que `range(n)` recorra "hasta n" en vez de "hasta n-1". Compruébalo siempre con un caso pequeño antes de confiar en el resultado con datos reales.

## 5. `enumerate()`: recorrer y contar a la vez

Cuando necesitas el índice y el valor a la vez, la forma manual (llevar un contador aparte) funciona, pero `enumerate()` es la forma idiomática y más difícil de desajustar por error.

````python
cartera = ["P001", "P002", "P003", "P004"]

# Forma manual (funciona, pero hay que acordarse de incrementar "i" cada vuelta)
i = 0
for id_poliza in cartera:
    print(f"Posición {i}: {id_poliza}")
    i += 1
````

````python
# Con enumerate(): mismo resultado, sin gestionar el contador a mano
for i, id_poliza in enumerate(cartera):
    print(f"Posición {i}: {id_poliza}")
````

````python
# enumerate() acepta un segundo argumento para empezar a contar donde quieras
# (útil, por ejemplo, para mostrar un "ranking" empezando en 1 en vez de en 0)
primas_ordenadas = [650.00, 480.50, 395.75]

for puesto, prima in enumerate(primas_ordenadas, start=1):
    print(f"{puesto}º mayor prima: {prima} €")
````

## 6. Funciones built-in de apoyo muy habituales con `for`

Python trae ya instaladas ("*built-in*") varias funciones que evitan escribir un bucle a mano para operaciones muy comunes sobre iterables.

````python
primas = [480.50, 210.00, 650.00, 395.75, 180.00]

print("Prima mínima: ", min(primas))
print("Prima máxima: ", max(primas))
print("Suma total:   ", sum(primas))
print("Prima media:  ", round(sum(primas) / len(primas), 2))  # round(valor, n_decimales)
````

````python
# abs(): valor absoluto -- típico al comparar una prima estimada con la real
prima_estimada = 500.0
prima_real = 480.50
desviacion = abs(prima_estimada - prima_real)
print(f"Desviación entre prima estimada y real: {desviacion:.2f} €")
````

````python
# zip(): recorre dos (o más) iterables EN PARALELO, emparejando elemento a elemento.
# Muy útil cuando tienes los mismos datos repartidos en listas separadas
# (algo habitual si vienen de columnas distintas de una hoja de cálculo).
ids_poliza = ["P001", "P002", "P003"]
primas_poliza = [480.50, 210.00, 650.00]

for id_poliza, prima in zip(ids_poliza, primas_poliza):
    print(f"{id_poliza}: {prima} €")

# list(zip(...)) construye la lista de tuplas de golpe, sin bucle explícito
print(list(zip(ids_poliza, primas_poliza)))
````

````python
# reversed(): recorre un iterable en orden inverso, sin modificar el original
for id_poliza in reversed(ids_poliza):
    print(id_poliza)

print("Original sin modificar:", ids_poliza)
````

### El módulo `random`

`random` no es una función built-in (hay que importarla), pero se usa constantemente junto a `for` para generar datos de prueba o simular escenarios — por ejemplo, simular la severidad de una muestra de siniestros.

````python
import random

random.seed(42)  # fija la "semilla": con la misma semilla, los resultados aleatorios son reproducibles

importes_simulados = [round(random.uniform(100, 20_000), 2) for _ in range(5)]
print("Importes de siniestros simulados:", importes_simulados)

print("El más caro de la muestra:", max(importes_simulados))
````

`random.seed(...)` es importante en un contexto de formación (y en general, siempre que necesites que un resultado "aleatorio" sea reproducible para poder comparar o depurar): sin fijarla, cada ejecución daría números distintos y sería imposible comparar resultados con un compañero o con la solución del ejercicio.

## Resumen

- `for elemento in <coleccion>:` es iteración definida: recorre, en orden, todos los elementos de un iterable.
- Un iterable produce un iterador con `iter()`; `next()` obtiene el siguiente valor. Cuando se agota, lanza `StopIteration` — `for` gestiona esto automáticamente por ti.
- Iterar un diccionario recorre sus llaves por defecto; usa `.values()` o `.items()` (con tuple unpacking `for k, v in d.items()`) según lo que necesites.
- `range(inicio, fin, paso)` genera enteros sin incluir `fin` — el error de "un elemento de más o de menos" por olvidar esto es muy común y no lanza ningún traceback.
- `enumerate(iterable, start=0)` da índice y valor a la vez, sin llevar un contador manual propenso a desajustarse.
- `min`, `max`, `sum`, `round`, `abs` cubren la mayoría de agregaciones simples sobre una lista de números sin necesidad de un bucle explícito.
- `zip()` recorre varios iterables en paralelo emparejando posición a posición; `reversed()` recorre en orden inverso sin modificar el original.
- `random` (con `random.seed(...)` para que sea reproducible) es la herramienta habitual para generar datos de prueba o simular escenarios dentro de un `for`.
