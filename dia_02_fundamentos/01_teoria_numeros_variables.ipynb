## 1. Tipos numéricos

Python tiene dos tipos numéricos básicos que usarás constantemente:

- **`int`** (entero): números sin decimales, positivos o negativos. Ej: `1280`, `-4`.
- **`float`** (punto flotante): números con decimales, positivos o negativos. Ej: `452.30`, `-3.14`.

La función `type()` te dice de qué tipo es un valor.

```python
# Número de pólizas en cartera (un conteo -> entero) y prima media (un importe -> flotante)
num_polizas = 1280
prima_media = 452.30

print(num_polizas, type(num_polizas))
print(prima_media, type(prima_media))
```

> 📊 **En Excel esto sería...** el formato de celda "Número" (sin decimales) frente a "Moneda"/"Número" con decimales. La diferencia en Excel es solo visual (formato de la celda); en Python es el **tipo de dato en sí**, y algunas operaciones se comportan distinto según el tipo (lo verás en el operador `/` más abajo).

## 2. Aritmética básica

| Operador | Símbolo | Ejemplo |
|---|---|---|
| Suma | `+` | `380 + 45.20` |
| Resta | `-` | `500 - 120` |
| Multiplicación | `*` | `12 * 38.5` |
| División | `/` | `100 / 3` (siempre devuelve `float`) |
| División entera | `//` | `100 // 3` (descarta el resto) |
| Módulo (resto) | `%` | `100 % 3` (solo el resto) |
| Potencia / raíz | `**` | `2 ** 3`, `9 ** (1/2)` |

```python
# Prima neta + recargo de gestión = prima total
prima_neta = 380
recargo_gestion = 45.20

prima_total = prima_neta + recargo_gestion
print("Prima total:", prima_total)
```

```python
# La division "/" siempre da float, aunque los dos operandos sean enteros
polizas_activas = 100
grupos_de_reparto = 3

print(polizas_activas / grupos_de_reparto)    # reparto exacto (float)
print(polizas_activas // grupos_de_reparto)   # cuantas polizas completas caben en cada grupo
print(polizas_activas % grupos_de_reparto)    # cuantas polizas "sobran" sin encajar en un grupo completo
```

`//` y `%` son muy útiles para repartos exactos: si quieres repartir 100 pólizas entre 3 gestores a partes iguales, `//` te dice cuántas le tocan a cada uno de forma exacta, y `%` cuántas quedan sueltas para repartir aparte.

```python
# Potencia y raiz cuadrada (una raiz cuadrada es una potencia de 1/2)
capital_inicial = 10_000
factor_revalorizacion = 1.03  # 3% anual

capital_a_3_anios = capital_inicial * factor_revalorizacion ** 3
print(round(capital_a_3_anios, 2))

raiz_cuadrada_de_9 = 9 ** (1 / 2)
print(raiz_cuadrada_de_9)
```

> ⚠️ **Error típico**: mezclar `/` cuando querías `//`, o al revés. `salario / 12` reparte el salario en 12 partes con decimales (correcto para una mensualidad); `salario // 12` lo trunca a un entero, perdiendo los céntimos — un error habitual si copias una fórmula pensada para conteos y la aplicas a importes monetarios.

## 3. Precedencia de operadores

Python respeta el orden matemático habitual: potencias antes que multiplicación/división, y estas antes que suma/resta. Los paréntesis `()` siempre tienen prioridad y son la forma de dejar explícito el orden que quieres, en vez de fiarte de memoria.

```python
# La tasa de IPT (Impuesto sobre Primas de Seguro) en Espana es del 6%
prima_base = 500

# Version SIN parentesis: por precedencia, "*" se calcula antes que "+",
# así que esto SÍ hace lo que parece (prima + el 6% de la prima)
prima_con_impuesto = prima_base + prima_base * 0.06
print("Con precedencia correcta:", prima_con_impuesto)

# Pero si la formula es mas larga, apoyarse solo en la precedencia es fragil.
# Ejemplo de bug real: calcular la prima media de dos recibos "a mano"
recibo_1 = 300
recibo_2 = 420

prima_media_mal = recibo_1 + recibo_2 / 2   # BUG: solo divide recibo_2 entre 2
prima_media_bien = (recibo_1 + recibo_2) / 2  # correcto: la suma completa se divide entre 2

print("Mal (sin parentesis):", prima_media_mal)
print("Bien (con parentesis):", prima_media_bien)
```

> ⚠️ **Error típico**: `recibo_1 + recibo_2 / 2` no calcula "la media de los dos recibos", porque `/` se evalúa antes que `+`. Es un error silencioso: el código no falla, simplemente calcula otra cosa. Cuando una fórmula tenga más de un operador, **añade paréntesis aunque no sean estrictamente necesarios** — cuesta un carácter y evita este tipo de bug.

**Para ti:** una póliza de hogar tiene un capital asegurado de `180_000`, una franquicia de `300` y una prima neta de `210.50`. Calcula el "capital neto de franquicia" (`capital asegurado - franquicia`) en una celda, usando paréntesis aunque la operación sea simple.

## 4. Asignación de variables: referencias, no cajas

El operador `=` no "guarda una copia" del valor dentro de la variable, como podrías imaginar de otros lenguajes. En Python, una variable es un **nombre que apunta (referencia) a un objeto** en memoria. La función `id()` te muestra esa referencia.

```python
x = 5
print("id(x):", id(x))
print("id(5):", id(5))  # el mismo id: x y el literal 5 apuntan al mismo objeto en memoria

x = 10
print("id(x) tras reasignar:", id(x))  # ahora apunta a otro objeto distinto
```

> 📊 **En Excel esto sería...** parecido a un *nombre definido* (`Fórmulas → Administrador de nombres`) que apunta a una celda concreta. Si rediriges ese nombre para que apunte a otra celda, el nombre "salta" a la nueva celda — no copia el valor viejo a ningún sitio, simplemente cambia a qué apunta. Una variable de Python se comporta igual: reasignar no modifica el objeto anterior, hace que el nombre apunte a uno nuevo.

## 5. Nombres de variables (PEP 8)

Reglas de [PEP 8](https://www.python.org/dev/peps/pep-0008/), la guía de estilo oficial de Python:

1. No pueden empezar por un número.
2. No pueden contener espacios — usa `_` para separar palabras (`snake_case`).
3. No pueden usar símbolos como `'",<>/?|\()!@#$%^&*~-+`.
4. Se usan minúsculas por convención (mayúsculas se reservan para clases y constantes).
5. Evita nombres de una sola letra, y sobre todo `l`, `O`, `I` (se confunden con `1` y `0`).
6. Evita nombres de palabras reservadas de Python (`list`, `str`, `type`, ...).

Ser explícito en los nombres no es un capricho: dentro de tres meses, `prima_riesgo` se entiende sin releer el código; `pr` o `x2` no.

```python
# Nombre de variable invalido: no puede empezar por un numero.
# Lo ejecutamos con exec() dentro de un try/except para ver el error sin romper el notebook.
codigo_invalido = "1_prima = 100"

try:
    exec(codigo_invalido)
except SyntaxError as e:
    print("Esto pasa si el nombre de variable empieza por un numero:")
    print(type(e).__name__, "-", e)
```

```python
# Ejemplo con nombres explicitos: prima de riesgo = frecuencia esperada de siniestro * coste medio del siniestro
frecuencia_siniestral = 0.15   # 15% de probabilidad de siniestro en el año
costo_medio_siniestro = 300    # euros

prima_riesgo = frecuencia_siniestral * costo_medio_siniestro
print("Prima de riesgo:", prima_riesgo)
```

## Resumen

- Python distingue `int` (enteros) de `float` (decimales); `type()` te dice cuál es cada uno.
- `/` siempre devuelve `float`; `//` y `%` sirven para repartos exactos (cociente y resto).
- La precedencia de operadores sigue el orden matemático habitual, pero **añadir paréntesis explícitos evita bugs silenciosos** en fórmulas con más de un operador.
- `=` no copia un valor dentro de una "caja": crea una referencia de un nombre a un objeto. `id()` te muestra esa referencia.
- Reasignar una variable no modifica el objeto anterior — el nombre pasa a apuntar a uno nuevo.
- Los nombres de variable siguen PEP 8: `snake_case`, minúsculas, explícitos, nunca empezando por un número.
