## 1. `if`: tomar una decisión

Hasta ahora el código se ejecutaba de arriba abajo, línea a línea. La sentencia `if` es la primera herramienta para que el programa **decida** qué hacer según una condición.

```python
if <expresión>:
    <bloque a ejecutar si la expresión es verdadera>
```

`<expresión>` se evalúa en contexto booleano (`True`/`False`), y el bloque indentado debajo solo se ejecuta si el resultado es `True`.

> 📊 **En Excel esto sería...** la función `SI(condición; valor_si_verdadero)`. La diferencia es que en Python el `if` no devuelve un valor por sí mismo: ejecuta o no un bloque de código.

````python
# Cartera de pólizas de ejemplo que usaremos durante todo el día de hoy.
polizas = [
    {"id": "P001", "tipo": "auto", "prima_anual": 480.50, "siniestros": 0, "antiguedad_anios": 3},
    {"id": "P002", "tipo": "hogar", "prima_anual": 210.00, "siniestros": 2, "antiguedad_anios": 1},
    {"id": "P003", "tipo": "vida", "prima_anual": 650.00, "siniestros": 0, "antiguedad_anios": 7},
    {"id": "P004", "tipo": "auto", "prima_anual": 395.75, "siniestros": 1, "antiguedad_anios": 5},
    {"id": "P005", "tipo": "hogar", "prima_anual": 180.00, "siniestros": 0, "antiguedad_anios": 2},
]

poliza = polizas[1]  # P002

if poliza["siniestros"] > 0:
    print(f"La póliza {poliza['id']} tiene siniestros abiertos: revisar antes de renovar.")
````

## 2. `elif` y `else`: cubrir todas las alternativas

`else` cubre "todo lo demás". `elif` (else + if) permite encadenar varias condiciones excluyentes: Python evalúa cada una en orden y ejecuta el bloque de la **primera** que sea verdadera; el resto ni se comprueban.

```python
if <expresión_1>:
    <bloque 1>
elif <expresión_2>:
    <bloque 2>
else:
    <bloque 3>
```

Un caso típico: clasificar la severidad de un siniestro según su importe.

````python
def clasificar_severidad(importe_siniestro: float) -> str:
    if importe_siniestro >= 20_000:
        return "muy grave"
    elif importe_siniestro >= 5_000:
        return "grave"
    elif importe_siniestro > 0:
        return "leve"
    else:
        return "sin siniestro"


for importe in [0, 1200, 8500, 45000]:
    print(f"Importe {importe:>7} € -> {clasificar_severidad(importe)}")
````

> 📊 **En Excel esto sería...** una cadena de `SI(...; ...; SI(...; ...; SI(...)))` anidados. En Python, `elif` evita anidar un `if` dentro de cada `else`, lo que hace el código mucho más legible cuantas más alternativas haya.

## 3. Combinar condiciones: `and`, `or`, `not`, y *short-circuit*

Las condiciones se pueden combinar con los operadores lógicos que ya conoces del Día 2. Python evalúa estas expresiones de izquierda a derecha y se detiene en cuanto el resultado ya está decidido — esto se llama ***short-circuit evaluation***.

Ejemplo: comprobar si una póliza es elegible para un descuento por baja siniestralidad (necesita antigüedad de al menos 2 años **y** cero siniestros).

````python
def es_elegible_descuento(poliza: dict) -> bool:
    return poliza["antiguedad_anios"] >= 2 and poliza["siniestros"] == 0


for p in polizas:
    print(f"{p['id']}: elegible = {es_elegible_descuento(p)}")
````

````python
# El short-circuit no es solo una curiosidad: evita errores en tiempo de ejecución.
# Si "prima_anual" pudiera faltar (dict incompleto), comprobamos primero que existe la llave
# antes de acceder a ella -- si la primera condición ya es False, Python nunca evalúa la segunda.
poliza_incompleta = {"id": "P006", "tipo": "auto"}  # sin "prima_anual"

if "prima_anual" in poliza_incompleta and poliza_incompleta["prima_anual"] > 1000:
    print("Prima alta")
else:
    print(f"{poliza_incompleta['id']}: no se puede evaluar la prima (dato ausente), se omite el chequeo.")
````

## 4. Expresiones condicionales (operador ternario)

Cuando lo único que hace el `if`/`else` es **asignar un valor**, se puede escribir en una sola línea:

```python
<valor_si_verdadero> if <expresión> else <valor_si_falso>
```

No es una estructura de control (no dirige el flujo del programa): es una expresión, como `2 + 2`, que se puede usar en cualquier sitio donde se espera un valor.

````python
# Forma con if/else tradicional
prima = 480.50
if prima > 400:
    categoria = "prima alta"
else:
    categoria = "prima estándar"
print(categoria)

# La misma lógica como expresión condicional (más compacto para asignaciones simples)
categoria = "prima alta" if prima > 400 else "prima estándar"
print(categoria)
````

````python
# Ejemplo aplicado: marcar cada póliza con un recargo si tiene más de un siniestro
for p in polizas:
    recargo = "con recargo" if p["siniestros"] > 1 else "sin recargo"
    print(f"{p['id']} ({p['tipo']}): {recargo}")
````

## 5. *Boolean cast*: qué valores son "falsos" sin ser `False`

Cuando `if` evalúa una expresión que no es directamente `True`/`False`, Python la convierte implícitamente a booleano (*boolean cast*). La regla general: **`0`, `0.0`, `""`, `[]`, `{}`, `()` y `None` se consideran `False`**; casi todo lo demás se considera `True`.

````python
valores_a_probar = {
    "cadena vacía": "",
    "cadena con texto": "auto",
    "cero": 0,
    "cero coma cero": 0.0,
    "número negativo": -2.1,
    "lista vacía": [],
    "diccionario con datos": {"id": "P001"},
    "None": None,
}

for etiqueta, valor in valores_a_probar.items():
    resultado = "VERDADERO" if valor else "FALSO"
    print(f"{etiqueta:>22}: {valor!r:>15} -> {resultado}")
````

Esto tiene una aplicación práctica muy común: comprobar si una lista de siniestros está vacía sin necesidad de usar `len(...) > 0`.

````python
siniestros_poliza_p001 = []  # sin siniestros registrados este año

if siniestros_poliza_p001:
    print(f"Hay {len(siniestros_poliza_p001)} siniestro(s) que revisar.")
else:
    print("Sin siniestros que revisar.")
````

> ⚠️ **Error típico: `IndentationError`**
>
> Python usa la indentación (espacios al principio de la línea) para saber qué código pertenece a qué bloque — no usa llaves `{}` como otros lenguajes. Si el bloque de un `if` no está indentado, o mezcla indentaciones distintas, el intérprete no sabe dónde empieza el bloque y lanza `IndentationError` **antes incluso de ejecutar nada** (es un error que se detecta al leer el código, no al ejecutarlo).
>
> Vamos a provocarlo a propósito, con `exec()`, para ver el traceback real sin que se rompa este notebook.

````python
codigo_con_error = '''
prima = 500
if prima > 400:
print("Prima alta")
'''

try:
    exec(codigo_con_error)
except IndentationError as e:
    print(f"{type(e).__name__}: {e}")
````

Cómo leer este traceback: `IndentationError: expected an indented block after 'if' statement on line 3` te dice exactamente qué esperaba Python (un bloque indentado) y dónde (línea 3, justo después del `if`). La corrección es añadir la indentación (normalmente 4 espacios) a la línea `print(...)`.

> ⚠️ **Error típico: `SyntaxError` por un `if` mal cerrado**
>
> El segundo error más común al empezar: olvidar los dos puntos (`:`) al final de la línea del `if`. Sin ellos, Python no reconoce dónde termina la condición y empieza el bloque.

````python
codigo_con_error_2 = '''
prima = 500
if prima > 400
    print("Prima alta")
'''

try:
    exec(codigo_con_error_2)
except SyntaxError as e:
    print(f"{type(e).__name__}: {e}")
````

`SyntaxError: expected ':'` es literal: falta el `:` al final de la línea del `if`. Este tipo de error siempre señala una línea y, a menudo, hasta una columna concreta — es la pista más directa que te da Python, y conviene acostumbrarse a leerla en vez de asumir "algo está mal en todo el bloque".

## Resumen

- `if <expr>:` ejecuta un bloque solo si `<expr>` es verdadera; `elif` encadena alternativas y `else` cubre el resto.
- `elif` evita anidar `if` dentro de cada `else` — mucho más legible con varias alternativas excluyentes.
- `and`, `or`, `not` combinan condiciones; Python usa *short-circuit*: en `A and B`, si `A` es `False`, `B` nunca se evalúa (útil para evitar errores, como comprobar que existe una llave antes de acceder a ella).
- La expresión condicional (`valor1 if expr else valor2`) es una forma compacta de asignar un valor según una condición — no sustituye a un `if` con varias sentencias dentro.
- El *boolean cast* convierte automáticamente cualquier valor a `True`/`False` en un `if`: `0`, `0.0`, `""`, `[]`, `{}`, `()` y `None` son "falsy"; casi todo lo demás es "truthy".
- `IndentationError` significa que Python esperaba un bloque indentado y no lo encontró; `SyntaxError` normalmente señala algo estructural (como un `:` que falta) — ambos se detectan **antes** de ejecutar el código, y el mensaje casi siempre dice la línea exacta.
- Leer el traceback de arriba abajo, fijándote en el tipo de error y el número de línea, es más rápido que releer todo el bloque a ciegas.
