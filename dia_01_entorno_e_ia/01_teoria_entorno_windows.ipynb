## 1. La terminal, tu nueva barra de fórmulas

> 📊 **En Excel esto sería...** la barra de fórmulas: en vez de hacer clic en menús para cada acción, escribes una instrucción directa y se ejecuta. La terminal (o línea de comandos) es lo mismo pero para todo el sistema, no solo para una celda.

En Windows tienes dos terminales instaladas de serie: **CMD** (`cmd.exe`) y **PowerShell**. Usaremos PowerShell, que es más moderna. La abres buscando "PowerShell" en el menú de inicio, o con `Win + X` → "Windows PowerShell" / "Terminal".

Comandos que vas a usar constantemente:

```powershell
cd ruta\a\una\carpeta   # moverte a esa carpeta (Change Directory)
cd ..                     # subir un nivel
dir                       # listar el contenido de la carpeta actual (equivale a "ls" en Mac/Linux)
mkdir nombre              # crear una carpeta
python --version          # comprobar la versión de Python instalada
```

`cd` sin nada te dice dónde estás; `dir` te dice qué hay ahí. Con esos dos te orientas en cualquier carpeta.

> ⚠️ **Error típico**: escribes `python` y la terminal responde `'python' no se reconoce como un comando interno o externo, programa o archivo por lotes ejecutable`. Significa que Windows no encuentra el programa `python.exe` porque no está en el `PATH` (la lista de carpetas donde el sistema busca programas). Revisa el paso 1 de `docs/guia_alumno_setup.md`: casi siempre se soluciona reinstalando Python marcando "Add python.exe to PATH", o abriendo una terminal nueva si acabas de instalarlo.

````python
# Python también puede consultar dónde está y qué hay alrededor, sin salir del notebook.
# pathlib es la forma moderna de trabajar con rutas de archivos en Python (mejor que concatenar texto).
from pathlib import Path

carpeta_actual = Path.cwd()
print("Estás ejecutando este notebook desde:", carpeta_actual)
````

````python
# Listar el contenido de la carpeta del curso (equivalente en Python al "dir" de la terminal)
raiz_curso = carpeta_actual.parent  # la carpeta padre de dia_01_entorno_e_ia/
elementos = sorted(p.name for p in raiz_curso.iterdir())
for nombre in elementos:
    print(nombre)
````

**Para ti:** antes de seguir, abre una terminal de verdad (no este notebook) en la carpeta del curso y ejecuta `dir`. Compara lo que ves con la salida de la celda anterior — debería coincidir. Es la misma información, dos formas de pedirla.

## 2. Entornos virtuales: por qué no instalamos nada "a pelo"

Cada proyecto de Python puede necesitar versiones distintas de las mismas librerías. Si instalas todo directamente en tu sistema, un proyecto puede romper a otro sin que lo veas venir.

Un **entorno virtual** es una copia aislada de Python con sus propias librerías, exclusiva para un proyecto. Es el equivalente a no mezclar las macros de un libro de Excel con las de otro: cada proyecto lleva las suyas.

```powershell
python -m venv .venv              # crear el entorno virtual (una vez por proyecto)
.venv\Scripts\activate            # activarlo (cada vez que abras una terminal nueva)
pip install -r requirements.txt   # instalar las dependencias del proyecto en ese entorno
```

Cuando el entorno está activo, verás `(.venv)` al principio de la línea de comandos. `requirements.txt`, en la raíz de este repositorio, es la lista de librerías que necesita el curso con su versión exacta — así todo el mundo ejecuta el mismo código con las mismas versiones.

````python
# Podemos inspeccionar requirements.txt como el archivo de texto que es
requisitos = (raiz_curso / "requirements.txt").read_text(encoding="utf-8")
lineas_paquetes = [l for l in requisitos.splitlines() if l and not l.startswith("#")]
print(f"Este curso fija {len(lineas_paquetes)} paquetes. Los primeros:")
for linea in lineas_paquetes[:5]:
    print(" -", linea)
````

## 3. Ejecutar código: notebook vs. script

En este curso trabajarás sobre todo con **notebooks** (`.ipynb`), como este mismo archivo: celdas que alternan explicación (markdown) y código, que se ejecutan una a una y recuerdan lo que ha pasado antes (las variables quedan "vivas" entre celdas). Es ideal para explorar, ver resultados intermedios y aprender.

Un **script** (`.py`) es un archivo de código plano que se ejecuta de arriba abajo de una vez, sin resultados intermedios visibles salvo que los imprimas tú. Es lo que usarás cuando el código ya está maduro y quieres automatizarlo (por ejemplo, los scripts de `data/generators/`).

```powershell
python mi_script.py       # ejecutar un script desde la terminal
```

Dentro de un notebook, `Shift + Enter` ejecuta la celda actual y pasa a la siguiente; `Ctrl + Enter` la ejecuta sin moverte. Si algo se queda raro (variables que no cuadran con lo que ves escrito), usa el menú **Kernel → Restart & Run All**: relanza el notebook de cero, de arriba abajo. Es tu botón de "recalcular todo" cuando algo no cuadra — el equivalente a recalcular una hoja de Excel entera.

> ⚠️ **Error típico**: ejecutas una celda y da un `NameError: name 'x' is not defined`, pero tú "juraría que ya lo definiste". Casi siempre es porque ejecutaste las celdas en un orden distinto al que aparecen (por ejemplo, saltaste una hacia arriba), o reiniciaste el kernel y no has vuelto a ejecutar todo desde el principio. La solución casi universal: **Kernel → Restart & Run All**.
