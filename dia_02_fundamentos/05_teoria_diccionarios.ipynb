## 1. Qué es un diccionario

Un **diccionario** (`dict`) es una colección de pares **clave-valor**: cada clave (única) tiene un valor asociado. A diferencia de una lista, no accedes por posición numérica, sino por clave.

Se crea con llaves `{}`, con la sintaxis `clave: valor` separada por comas.

```python
poliza = {
    "poliza_id": "AUTO-2026-00147",
    "tomador": "García, S.",
    "prima": 480.5,
    "activa": True,
}
print(poliza)
```

> 📊 **En Excel esto sería...** una tabla de búsqueda usada con `BUSCARV()`/`VLOOKUP`: la primera columna son las claves (algo único, como un DNI o un `poliza_id`), y cada columna siguiente es un valor asociado a esa clave. `poliza["prima"]` es como hacer `=BUSCARV("AUTO-2026-00147"; tabla; columna_prima; FALSO)` — pero sin tener que buscar en toda la tabla: el diccionario "sabe" ir directo a la clave.

```python
# Un diccionario puede construirse tambien con la funcion dict()
franquicias = dict([("AUTO", 300), ("HOGAR", 150), ("VIDA", 0)])
print(franquicias)

# Y con "dict comprehension": crear un diccionario a partir de una expresion
cuadrados = {n: n ** 2 for n in [1, 2, 3, 4]}
print(cuadrados)
```

## 2. Acceso a los valores

```python
poliza = {"poliza_id": "AUTO-2026-00147", "tomador": "García, S.", "prima": 480.5}

print(poliza["prima"])            # acceso directo por clave
print(poliza.keys())              # todas las claves
print(poliza.values())            # todos los valores

poliza["prima"] = poliza["prima"] + 20   # se puede reasignar el valor de una clave existente
print(poliza)

poliza["activa"] = True   # y crear una clave nueva simplemente asignandole un valor
print(poliza)
```

```python
# Que pasa si accedes a una clave que no existe?
try:
    poliza["email"]
except KeyError as e:
    print("Esto pasa si accedes a una clave que no existe:")
    print(type(e).__name__, "-", e)
```

> ⚠️ **Error típico**: el `KeyError` de la celda anterior aparece cuando accedes con `[...]` a una clave que no existe. Si no estás seguro de que la clave exista (por ejemplo, un campo opcional que no todas las pólizas tienen), usa `.get("clave", valor_por_defecto)` en vez de `[...]` — no lanza error, devuelve el valor por defecto que tú decidas.

```python
print(poliza.get("email", "sin email registrado"))   # no falla, devuelve el valor por defecto
print(poliza.get("prima", 0))                            # si la clave SI existe, devuelve su valor normal
```

## 3. Diccionarios anidados

Un valor puede ser, a su vez, otro diccionario (o una lista). Es la forma natural de representar datos con estructura, como una póliza con su tomador y sus coberturas.

```python
poliza = {
    "poliza_id": "AUTO-2026-00147",
    "tomador": {
        "nombre": "García, S.",
        "dni": "12345678A",
    },
    "coberturas": {
        "responsabilidad_civil": True,
        "lunas": True,
        "asistencia_en_viaje": False,
    },
}

print(poliza["tomador"])                      # el diccionario anidado completo
print(poliza["tomador"]["nombre"])             # un valor dentro del diccionario anidado
print(poliza["coberturas"]["lunas"])           # otra rama del anidamiento
```

## 4. Recorrer un diccionario: `.items()`

```python
franquicias = {"AUTO": 300, "HOGAR": 150, "VIDA": 0}

for ramo, importe in franquicias.items():
    print(f"Ramo {ramo}: franquicia de {importe} EUR")
```

```python
# .items() no devuelve una lista, devuelve un objeto "vista" -- no se puede indexar directamente
try:
    franquicias.items()[0]
except TypeError as e:
    print("Esto pasa si intentas indexar .items() como si fuera una lista:")
    print(type(e).__name__, "-", e)
```

> ⚠️ **Error típico**: el `TypeError` de la celda anterior es intencionado — es un buen ejemplo de un error real que te vas a encontrar. `.items()` devuelve un objeto `dict_items`, pensado para **iterar** (recorrer con `for`), no para indexar por posición como una lista. Si de verdad necesitas el primer par clave-valor por posición, tienes que convertirlo explícitamente a lista primero: `list(franquicias.items())[0]`.

```python
print(list(franquicias.items())[0])   # ahora si funciona, porque list() lo convierte a una lista de tuplas
```

## 5. `dict comprehension` y otros métodos

```python
primas_por_poliza = {"P001": 420, "P002": 380, "P003": 510, "P004": 275}

# dict comprehension con condicion: solo las polizas con prima superior a 400
primas_altas = {k: v for k, v in primas_por_poliza.items() if v > 400}
print(primas_altas)

# .pop() elimina una clave y devuelve su valor
prima_eliminada = primas_por_poliza.pop("P004")
print("Eliminada:", prima_eliminada, "-- diccionario restante:", primas_por_poliza)
```

**Para ti:** dado `primas_por_poliza = {"P001": 420, "P002": 380, "P003": 510}`, construye con un *dict comprehension* un nuevo diccionario `primas_con_ipt` que aplique un 6% de IPT a cada prima (redondeado a 2 decimales con `round(valor, 2)`).

```python
primas_por_poliza = {"P001": 420, "P002": 380, "P003": 510}

# tu codigo aqui
primas_con_ipt = {k: round(v * 1.06, 2) for k, v in primas_por_poliza.items()}
print(primas_con_ipt)
```

## Resumen

- Un diccionario (`dict`) asocia claves únicas a valores; se accede por clave, no por posición.
- `diccionario["clave"]` lanza `KeyError` si la clave no existe; `.get("clave", valor_por_defecto)` no falla nunca.
- Un valor puede ser, a su vez, otro diccionario o una lista (diccionarios anidados).
- `.items()` es para iterar con `for clave, valor in dict.items():`, no para indexar por posición — necesitas `list(...)` primero si quieres eso.
- Un *dict comprehension* (`{k: expr for k, v in dict.items() if condicion}`) construye un diccionario nuevo en una línea.
- Un diccionario es, conceptualmente, una tabla de búsqueda tipo `BUSCARV` — pero con una diferencia real: no admite claves repetidas.
- Pedirle a la IA una analogía de Excel es útil para fijar un concepto nuevo, pero la analogía se valida ejecutando código y buscando los casos donde deja de cumplirse — no basta con que "suene bien".
