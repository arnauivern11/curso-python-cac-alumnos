## 1. Qué es un string y cómo se crea

Un **string** (`str`) es una secuencia de caracteres: texto. Se crea con comillas dobles `"..."` o simples `'...'` — da igual cuál uses, mientras seas consistente. Si el propio texto contiene comillas dobles, usa comillas simples por fuera (o viceversa).

```python
producto = "Auto"
compania = 'Colegio de Actuarios de Cataluña'
observacion = 'El asegurado indica: "no hubo testigos" en el parte.'

print(producto)
print(compania)
print(observacion)
```

Si escribes un string como última línea de una celda, Jupyter lo muestra automáticamente — pero **solo el último**. Para mostrar varios valores, o para verlos desde un script (no un notebook), usa siempre `print()`.

```python
"esto no se ve"
"esto tampoco"
print("esto si se ve")
print("y esto tambien")
```

Dos caracteres especiales que verás constantemente dentro de un string: `\n` (salto de línea) y `\t` (tabulación, ~4 espacios).

```python
print("Póliza: P00123\nTomador: García, S.\nPrima: 480.00 EUR")
print("Ramo:\tAUTO")
print("Ramo:\tHOGAR")
```

## 2. Indexado (*indexing*) y troceado (*slicing*)

Un string es una **secuencia**: cada carácter tiene una posición (índice), empezando en `0`. Puedes acceder a un carácter con `texto[posicion]`, y a un tramo con `texto[inicio:fin]` (el índice `fin` **no se incluye**).

```python
poliza_id = "AUTO-2026-00147"

print(poliza_id[0])     # primer caracter
print(poliza_id[-1])    # ultimo caracter (indices negativos cuentan desde el final)
print(poliza_id[0:4])   # el ramo: desde el indice 0 (incluido) hasta el 4 (excluido)
print(poliza_id[5:9])   # el anio
print(poliza_id[-5:])   # los ultimos 5 caracteres: el numero de poliza
```

> 📊 **En Excel esto sería...** las funciones `IZQUIERDA()`/`LEFT()`, `DERECHA()`/`RIGHT()` y `EXTRAE()`/`MID()`: `poliza_id[0:4]` es como `=IZQUIERDA(A2;4)`, y `poliza_id[-5:]` es como `=DERECHA(A2;5)`. El slicing de Python es más flexible porque combina ambas ideas (inicio y fin) en una sola sintaxis.

```python
# El tercer numero opcional en el slice es el "paso" (step)
alfabeto_ramos = "AUTOHOGARVIDA"
print(alfabeto_ramos[::2])    # un caracter de cada dos
print(alfabeto_ramos[::-1])   # el string invertido -- paso -1 recorre de atras hacia adelante
```

**Para ti:** dado `poliza_id = "HOGAR-2025-00892"`, extrae por separado el ramo (`"HOGAR"`), el año (`"2025"`) y el número de póliza (`"00892"`) usando solo slicing (sin métodos todavía).

```python
poliza_id = "HOGAR-2025-00892"

# tu codigo aqui
ramo = poliza_id[0:5]
anio = poliza_id[6:10]
numero = poliza_id[-5:]

print(ramo, anio, numero)
```

## 3. Inmutabilidad

Un string **no se puede modificar in-place**: una vez creado, sus caracteres no cambian. Cualquier operación que "parece" modificarlo en realidad crea un string nuevo.

```python
poliza_id = "AUTO-2026-00147"

try:
    poliza_id[0] = "a"
except TypeError as e:
    print("Esto pasa si intentas modificar un caracter de un string:")
    print(type(e).__name__, "-", e)
```

> ⚠️ **Error típico**: el `TypeError: 'str' object does not support item assignment` de la celda anterior es la forma que tiene Python de decirte "los strings son inmutables, no puedes cambiar un carácter en su sitio". La solución casi siempre es construir un string **nuevo** a partir del original (con slicing, concatenación o un método), y guardarlo en la misma variable o en una nueva — nunca "editarlo por dentro".

## 4. Operadores con strings: `+` y `*`

Los strings soportan `+` (concatenar) y `*` (repetir).

```python
tomador = "García, S."
ramo = "AUTO"

descripcion = ramo + " - " + tomador   # "+" crea un string NUEVO, no modifica los originales
print(descripcion)

separador = "-" * 20
print(separador)
```

## 5. Métodos de string

Los métodos son acciones ya construidas que puedes aplicar a un string con `texto.metodo()`. Aquí los más habituales — la [lista completa está en la documentación oficial](https://docs.python.org/3/library/stdtypes.html#string-methods).

```python
observacion = "  Siniestro con Daños en Parabrisas  "

print(observacion.upper())               # todo mayusculas
print(observacion.lower())               # todo minusculas
print(observacion.strip())               # quita espacios sobrantes al inicio/final
print(observacion.strip().capitalize())  # quita espacios y pone formato "Nombre propio"
```

```python
texto_parte = "Colisión con daño en parachoques y daño en faro"

print(texto_parte.split())          # separa por espacios (por defecto) -> lista de palabras
print(texto_parte.split("daño"))    # separa por el texto indicado
print(texto_parte.count("daño"))    # cuantas veces aparece "daño" (distingue mayus/minus)
print(texto_parte.lower().count("daño"))  # por eso conviene normalizar a minusculas antes de contar
print(texto_parte.find("daño"))     # posicion de la PRIMERA aparicion
print(texto_parte.replace("daño", "avería"))  # sustituye todas las apariciones
```

> ⚠️ **Error típico**: Python distingue mayúsculas de minúsculas (`"Daño" != "daño"`). Si vas a contar o comparar texto que puede venir con capitalización distinta (por ejemplo, partes de siniestro escritos por personas distintas), normaliza siempre primero con `.lower()` o `.upper()` antes de comparar o contar — si no, subestimarás el resultado sin ningún error visible.

**Para ti:** dado `parte = "  el asegurado declara DAÑOS en el retrovisor  "`, limpia los espacios sobrantes, ponlo en formato de frase (mayúscula solo al inicio) y cuenta cuántas veces aparece la letra `"a"` ignorando mayúsculas/minúsculas.

```python
parte = "  el asegurado declara DAÑOS en el retrovisor  "

# tu codigo aqui
parte_limpio = parte.strip().capitalize()
apariciones_de_a = parte.lower().count("a")

print(repr(parte_limpio))
print(apariciones_de_a)
```

## 6. Formateo: `.format()`, `%` y f-strings

Hay tres formas de insertar el valor de una variable dentro de un string. Las tres existen en código real, así que conviene reconocer las tres — pero en código nuevo, **usa siempre f-strings** (son las más claras y las más rápidas).

```python
prima = 452.317
ramo = "AUTO"

# Metodo .format() -- habitual en codigo mas antiguo
print("Ramo {} - prima {:.2f} EUR".format(ramo, prima))

# Metodo con marcadores % -- aun mas antiguo, evitalo en codigo nuevo
print("Ramo %s - prima %.2f EUR" % (ramo, prima))

# f-string -- la forma moderna y recomendada (Python 3.6+)
print(f"Ramo {ramo} - prima {prima:.2f} EUR")
```

> 📊 **En Excel esto sería...** la función `TEXTO()`/`TEXT()` combinada con `&` (o `CONCATENAR`): `f"prima {prima:.2f} EUR"` es el equivalente de `="prima " & TEXTO(B2;"0,00") & " EUR"` — el `:.2f` dentro de la f-string fija 2 decimales, igual que el formato `"0,00"` en `TEXTO()`.

```python
# f-strings tambien evaluan expresiones dentro de las llaves, no solo variables sueltas
prima_neta = 500
tasa_ipt = 0.06

print(f"Prima total con IPT: {prima_neta * (1 + tasa_ipt):.2f} EUR")
```

**Para ti:** con `poliza_id = "AUTO-2026-00147"`, `tomador = "García, S."` y `prima = 480.5`, construye con una f-string el mensaje exacto: `"Póliza AUTO-2026-00147 (García, S.): 480.50 EUR"` — fíjate en los dos decimales de la prima.

```python
poliza_id = "AUTO-2026-00147"
tomador = "García, S."
prima = 480.5

# tu codigo aqui
mensaje = f"Póliza {poliza_id} ({tomador}): {prima:.2f} EUR"
print(mensaje)
```

## Resumen

- Un string se crea con `"..."` o `'...'`; dentro de un notebook, solo se muestra automáticamente el último valor de la celda — usa `print()` para el resto.
- `texto[i]` accede a un carácter; `texto[inicio:fin]` trocea (el `fin` no se incluye); los índices negativos cuentan desde el final.
- Los strings son **inmutables**: `texto[0] = "x"` da `TypeError`. Cualquier "modificación" crea un string nuevo.
- `+` concatena, `*` repite.
- Métodos habituales: `.upper()`, `.lower()`, `.strip()`, `.split()`, `.count()`, `.find()`, `.replace()`, `.capitalize()`.
- Python distingue mayúsculas/minúsculas — normaliza con `.lower()`/`.upper()` antes de comparar o contar texto.
- Para formatear texto con variables, usa **f-strings** (`f"texto {variable:.2f}"`); `.format()` y `%` existen en código heredado pero no se usan en código nuevo.
