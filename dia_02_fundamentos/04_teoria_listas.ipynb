## 1. Qué es una lista

Una **lista** es una colección **ordenada** de elementos, que pueden ser de cualquier tipo (incluso mixtos). Se crea con corchetes `[]`, separando los elementos por comas.

```python
primas_cartera = [420, 380, 510, 275, 640]
print(primas_cartera)

# Puede mezclar tipos, aunque en la practica lo habitual es que todos los elementos sean del mismo tipo
poliza_mixta = ["AUTO-2026-00147", 480.5, True]
print(poliza_mixta)
```

> 📊 **En Excel esto sería...** una columna de celdas, o un rango de una sola fila/columna: `primas_cartera` es como seleccionar `A2:A6` con las primas de cinco pólizas. La diferencia es que en Python ese rango es un objeto que puedes guardar en una variable, pasar a una función y transformar con código.

```python
# range() genera una secuencia de numeros -- muy util para crear listas rapido
polizas_numeradas = list(range(1, 11))  # del 1 al 10
print(polizas_numeradas)
print("Numero de elementos:", len(polizas_numeradas))
```

## 2. Indexado y slicing

Funcionan **exactamente igual** que en los strings: índice `0` es el primero, índices negativos cuentan desde el final, y el slicing `[inicio:fin:paso]` extrae un tramo.

```python
primas_cartera = [420, 380, 510, 275, 640]

print(primas_cartera[0])     # primera prima
print(primas_cartera[-1])    # ultima prima
print(primas_cartera[1:4])   # de la segunda a la cuarta (indice 4 excluido)
print(primas_cartera[::2])   # una de cada dos
```

## 3. Mutabilidad: la gran diferencia con los strings

A diferencia de los strings, las listas **sí se pueden modificar in-place**: cambiar un elemento, añadir o quitar no crea una lista nueva, modifica la misma.

```python
primas_cartera = [420, 380, 510, 275, 640]
print("id antes:", id(primas_cartera))

primas_cartera[0] = 450   # esto SI funciona (a diferencia de un string)
print(primas_cartera)
print("id despues:", id(primas_cartera))  # el mismo id: es el MISMO objeto, modificado
```

Esto tiene una consecuencia importante: si asignas una lista existente a otra variable con `=`, **no la copias** — ambos nombres apuntan al mismo objeto. Modificar una "modifica" la otra, porque en realidad son la misma.

```python
primas_original = [420, 380, 510]
primas_alias = primas_original       # NO es una copia: mismo objeto con dos nombres

primas_alias[0] = 999
print("primas_alias:   ", primas_alias)
print("primas_original:", primas_original)   # tambien cambio, aunque no la tocaste "directamente"
print(primas_original is primas_alias)
```

> ⚠️ **Error típico**: `cartera_backup = cartera` seguido de modificar `cartera_backup` pensando que `cartera` queda intacta — no es así, porque `cartera_backup` no es una copia, es **el mismo objeto**. Si de verdad quieres una copia independiente, usa `.copy()` (o `list(cartera)`, o el slicing completo `cartera[:]`).

```python
primas_original = [420, 380, 510]
primas_copia = primas_original.copy()   # copia real: objeto nuevo con el mismo contenido

primas_copia[0] = 999
print("primas_copia:   ", primas_copia)
print("primas_original:", primas_original)   # esta vez SI queda intacta
print(primas_original is primas_copia)
```

**Para ti:** crea `siniestros_enero = [1200, 850, 0, 3400]`. Crea una variable `siniestros_backup` que sea una **copia real e independiente** (no un alias), cambia el primer elemento de `siniestros_enero` a `0`, y comprueba que `siniestros_backup` no se ha visto afectado.

```python
siniestros_enero = [1200, 850, 0, 3400]

# tu codigo aqui
siniestros_backup = siniestros_enero.copy()
siniestros_enero[0] = 0

print(siniestros_enero)
print(siniestros_backup)
```

## 4. Operadores `+` y `*`

Igual que con strings: `+` concatena dos listas (devuelve una lista nueva), `*` repite el contenido.

```python
primas_ramo_auto = [420, 380]
primas_ramo_hogar = [275, 640]

todas_las_primas = primas_ramo_auto + primas_ramo_hogar
print(todas_las_primas)

recordatorios = ["revisar"] * 3
print(recordatorios)
```

## 5. Métodos básicos

```python
primas = [420, 380, 510]

primas.append(600)         # añade UN elemento al final
print("append:", primas)

primas.extend([275, 640])  # añade CADA elemento de un iterable al final
print("extend:", primas)

primas.insert(1, 999)      # inserta en una posicion concreta
print("insert:", primas)
```

> ⚠️ **Error típico**: usar `.append()` cuando querías `.extend()`. `lista.append([1, 2])` no añade `1` y `2` por separado — añade **la lista `[1, 2]` entera como un único elemento**, generando una lista anidada por accidente. Si quieres "pegar" los elementos de un iterable al final, usa `.extend()`.

```python
primas = [420, 380]
primas.append([275, 640])   # ojo: esto NO hace lo que probablemente esperas
print(primas)                # [420, 380, [275, 640]]  <- lista anidada, no 4 elementos sueltos
```

```python
primas = [420, 380, 510, 275]

eliminada_por_valor = primas.remove(510)  # elimina la PRIMERA aparicion de ese valor (no devuelve nada)
print("tras remove(510):", primas)

valor_eliminado = primas.pop(0)   # elimina y DEVUELVE el elemento en esa posicion
print("valor eliminado con pop(0):", valor_eliminado)
print("lista tras pop:", primas)

primas.sort(reverse=True)   # ordena in-place, de mayor a menor
print("ordenada desc:", primas)
```

## 6. Listas anidadas

Una lista puede contener otras listas — es la forma más simple de representar una tabla (filas y columnas) sin usar todavía pandas.

```python
# Cada fila: [poliza_id, ramo, prima]
cartera = [
    ["P001", "AUTO", 420],
    ["P002", "HOGAR", 275],
    ["P003", "AUTO", 510],
]

print(cartera[0])       # primera fila completa
print(cartera[0][2])    # prima de la primera fila (fila 0, columna 2)
print(cartera[1][1])    # ramo de la segunda fila
```

**List comprehension**: una forma compacta de construir una lista nueva a partir de otra, iterando en una sola línea.

```python
# Extraer solo la columna de primas (columna 2) de toda la cartera
primas_de_la_cartera = [fila[2] for fila in cartera]
print(primas_de_la_cartera)

# Con condicion: solo las primas de polizas de AUTO
primas_de_auto = [fila[2] for fila in cartera if fila[1] == "AUTO"]
print(primas_de_auto)
```

> 📊 **En Excel esto sería...** una tabla dinámica o `FILTRAR()`/`FILTER()`: `[fila[2] for fila in cartera if fila[1] == "AUTO"]` es equivalente a filtrar la columna "Prima" donde la columna "Ramo" sea "AUTO". Es la misma idea de siempre — filtrar y quedarte con una columna — pero en una línea de código en vez de un menú.

**Para ti:** dada `siniestros = [1200, 0, 850, 0, 3400, 150]`, usa una list comprehension para quedarte solo con los siniestros mayores que `0` (los que realmente han costado dinero, descartando los meses sin siniestralidad).

```python
siniestros = [1200, 0, 850, 0, 3400, 150]

# tu codigo aqui
siniestros_con_coste = [s for s in siniestros if s > 0]
print(siniestros_con_coste)
```

## Resumen

- Una lista se crea con `[]`; admite indexado y slicing igual que un string, más `len()` para su tamaño.
- Las listas son **mutables**: se pueden modificar in-place, a diferencia de los strings.
- Asignar una lista a otra variable con `=` **no la copia** — crea un alias del mismo objeto. Para copiar de verdad, usa `.copy()`.
- `.append()` añade un único elemento (¡ojo si le pasas una lista, la anida!); `.extend()` añade cada elemento de un iterable por separado.
- `.insert()`, `.remove()`, `.pop()` y `.sort()` son los métodos de manipulación más habituales.
- Una lista puede contener otras listas (listas anidadas) — es la forma más simple de representar filas y columnas antes de llegar a pandas.
- Una *list comprehension* (`[expr for elemento in iterable if condicion]`) construye una lista nueva filtrando o transformando otra en una sola línea.
